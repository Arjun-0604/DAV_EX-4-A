```python


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm

# -----------------------------
# Create Sample UCI Diabetes Dataset
# -----------------------------
np.random.seed(42)

rows = 100

uci_diabetes = pd.DataFrame({
    "Glucose": np.random.randint(70, 200, rows),
    "BloodPressure": np.random.randint(50, 110, rows),
    "SkinThickness": np.random.randint(10, 50, rows),
    "Insulin": np.random.randint(20, 300, rows),
    "BMI": np.random.uniform(18, 45, rows),
    "DiabetesPedigreeFunction": np.random.uniform(0.1, 2.5, rows),
    "Age": np.random.randint(20, 80, rows),
    "Outcome": np.random.randint(0, 2, rows)
})

# Save dataset (Optional)
uci_diabetes.to_csv("uci_diabetes (3).csv", index=False)

print("========== UCI Diabetes Dataset Sample ==========")
print(uci_diabetes.head())

# -----------------------------
# Plot Normal Distribution Curves
# -----------------------------
plt.figure(figsize=(12,5))

# Glucose Distribution
plt.subplot(1,2,1)
sns.histplot(uci_diabetes["Glucose"], kde=True, stat="density", color="skyblue")

x = np.linspace(
    uci_diabetes["Glucose"].min(),
    uci_diabetes["Glucose"].max(),
    100
)

plt.plot(
    x,
    norm.pdf(
        x,
        uci_diabetes["Glucose"].mean(),
        uci_diabetes["Glucose"].std()
    ),
    color="red",
    linewidth=2
)

plt.title("Normal Curve - Glucose")
plt.xlabel("Glucose")
plt.ylabel("Density")

# BMI Distribution
plt.subplot(1,2,2)
sns.histplot(uci_diabetes["BMI"], kde=True, stat="density", color="lightgreen")

x = np.linspace(
    uci_diabetes["BMI"].min(),
    uci_diabetes["BMI"].max(),
    100
)

plt.plot(
    x,
    norm.pdf(
        x,
        uci_diabetes["BMI"].mean(),
        uci_diabetes["BMI"].std()
    ),
    color="red",
    linewidth=2
)

plt.title("Normal Curve - BMI")
plt.xlabel("BMI")
plt.ylabel("Density")

plt.tight_layout()
plt.show()

print("\nProgram Executed Successfully!")
```
